```python
# wap to read a no from the user and find the factorial using user defined function
n = int(input("Enter a number: "))

def factorial(n):
    fact = 1
    for i in range(1, n + 1):
        fact = fact * i
    return fact

print("Factorial is:", factorial(n))
```

```python
#create a class of circles and print area and perimeter

class Circle():
  def get_data(self):
    self.r=int(input("Enter the radius: "))

  def display_data(self):
    print(self.r)

  def area(self):
    print(3.14*self.r*self.r)
  def perimeter(self):
    print(2*3.14*self.r)

c1=Circle()

c1.get_data()
c1.display_data()
c1.area()
c1.perimeter()
```

```python
class Rectangle:
    def __init__(self, length, breadth):
        self.length = length
        self.breadth = breadth

    def display_data(self):
        print("Length:", self.length)
        print("Breadth:", self.breadth)

    def area(self):
        print("Area:", self.length * self.breadth)

    def perimeter(self):
        print("Perimeter:", 2 * (self.length + self.breadth))


r1 = Rectangle(12, 15)

r1.display_data()
r1.area()
r1.perimeter()
```

```python
#constructor



class Student:
    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno

    def display_data(self):
        print("Name:", self.name)
        print("Roll No:", self.rollno)


class Test(Student):
    def __init__(self, name, rollno, tmark):
        super().__init__(name, rollno)
        self.tmark = tmark


class Sports(Student):
    def __init__(self, name, rollno, smark):
        super().__init__(name, rollno)
        self.smark = smark


class Grade(Test, Sports):
    def __init__(self, name, rollno, tmark, smark):
        Student.__init__(self, name, rollno)
        self.tmark = tmark
        self.smark = smark

    def calgrade(self):
        total = self.tmark + self.smark
        print("Total Marks:", total)
        if total > 80:
            print("Grade A")
        elif total > 50:
            print("Grade B")
        else:
            print("Failed")


g = Grade("Abu", 42, 40, 25)
g.display_data()
g.calgrade()
```
